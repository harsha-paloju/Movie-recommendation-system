```python
import pandas as pd
```

```python
movies=pd.read_csv("tmdb_5000_movies.csv")
credits=pd.read_csv("tmdb_5000_credits.csv")
```

```python
credits
```

```python
movies=pd.merge(movies,credits,on="title")
```

```python
movies
```

```python
movies.columns
```

```python
movies=movies.drop(["budget","genres","homepage","original_title","popularity","production_companies","production_countries","revenue","runtime","status","tagline","vote_average","vote_count",],axis=1)
```

```python
movies=movies.drop(["movie_id"],axis=1)
```

```python
movies
```

```python
movies.isnull().sum()
```

```python
movies["overview"]=movies["overview"].fillna("")
```

```python
movies["release_date"]=movies["release_date"].fillna("")
```

```python
movies["release_date"]=pd.to_datetime(movies["release_date"])
```

```python
movies["year"]=movies["release_date"].dt.year
```

```python
movies=movies.drop(4559)
```

```python
movies.isnull().sum()
```

```python
movies=movies.drop("release_date",axis=1)
```

```python
movies
```

```python
import ast
movies["keywords"]=movies["keywords"].apply(lambda x:ast.literal_eval(x))
```

```python
def fun_for_keywords(x):
    l=[]
    for each in x:
        l.append(each["name"])
        
    return l

movies["keywords"]=movies["keywords"].apply(fun_for_keywords)
```

```python
movies
```

```python
movies["overview"]=movies["overview"].apply(lambda x:x.replace(",",""))
movies["overview"]=movies["overview"].apply(lambda x:x.replace(".",""))
```

```python
movies["overview"]=movies["overview"].apply(lambda x:x.split())
```

```python
movies["original_language"]=movies["original_language"].apply(lambda x:x.split())
```

```python
movies["spoken_languages"]=movies["spoken_languages"].apply(lambda x:ast.literal_eval(x))
```

```python
movies["spoken_languages"][0]
```

```python
def fun_for_spoken_languages(x):
    l=[]
    for each in x:
        l.append(each["name"])
        
    return l

movies["spoken_languages"]=movies["spoken_languages"].apply(fun_for_spoken_languages)
```

```python
movies
```

```python
movies["cast"]=movies["cast"].apply(lambda x:ast.literal_eval(x))
```

```python
movies["cast"][0]
```

```python
movies=movies.drop("cast",axis=1)
```

```python
movies["crew"]=movies["crew"].apply(lambda x:ast.literal_eval(x))
```

```python
def fun_for_crew(x):
    l=[]
    for each in x:
        if each["job"]=="Director" or each["job"]=="Writer" or each["job"]=="Screenplay":
            l.append(each["name"])
    return l

movies["crew"]=movies["crew"].apply(fun_for_crew)
            
```

```python
movies["crew"]=movies["crew"].apply(lambda x:list(set(x)))
```

```python
movies["crew"]
```

```python
movies["year"]=movies["year"].astype(int)
```

```python
movies["year"]=movies["year"].astype(str)
```

```python
def fun_for_year(x):
    l=[]
    l.append(x)
    return l

movies["year"]=movies["year"].apply(fun_for_year)
```

```python
movies
```

```python
movies.info()
```

```python
movies["tags"]=movies["keywords"]+movies["original_language"]+movies["overview"]+movies["spoken_languages"]+movies["crew"]+movies["year"]
```

```python
movies=movies[["id","title","tags"]]
```

```python
movies
```

```python
movies["tags"]=movies["tags"].apply(lambda x:",".join(x))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer(stop_words="english",max_features=9000)
vectors=vectorizer.fit_transform(movies["tags"])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarities=cosine_similarity(vectors)
```

```python
similarities.shape
```

```python
desire_id=3
recommendations=[]
for idx,similarity in sorted(enumerate(similarities[desire_id]),key=lambda x:x[1],reverse=True)[:10]:
    recommendations.append([idx,similarity])
    
```

```python
for idx,similarity in recommendations:
    if idx==desire_id:
        continue
    print(movies.iloc[idx]["title"])
```

```python
new_similarities=[list(row) for row in similarities]
```

```python
for row in new_similarities:
    for i in range(len(row)):
        row[i]=float(row[i])
    
```

```python
for row in new_similarities:
    for i in range(len(row)):
        row[i]=[i,row[i]]
```

```python
for i in range(len(new_similarities)):
    
    new_row=sorted(new_similarities[i],key=lambda x:x[1],reverse=True)[:15]
    
    new_similarities[i]=new_row
```

```python
for idx,score in new_similarities[20]:
    print(movies.iloc[idx]["title"])
```

```python
movies=movies.drop("tags",axis=1)
```

```python
movies
```

```python
import pickle

with open("recommendations.pkl","wb") as f:
    pickle.dump({"similarities":new_similarities,"movies":movies},f)
```
